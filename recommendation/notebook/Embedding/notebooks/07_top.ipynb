```python
import pandas as pd

# Đọc file output AI cuối
df = pd.read_excel("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/Embedding/data_outputs/14_candidate_job_explanations.xlsx")
# nếu bạn muốn dùng parquet thì:
# df = pd.read_parquet("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/Embedding/data_outputs/14_candidate_job_explanations.parquet")

# Lấy top 5 cho mỗi candidate
top5_df = (
    df.sort_values(["candidate_id", "hybrid_rank"])
      .groupby("candidate_id")
      .head(5)
      .copy()
)

# Sắp xếp lại cho đẹp
top5_df = top5_df.sort_values(["candidate_id", "hybrid_rank"]).reset_index(drop=True)

# Lưu file
top5_df.to_excel("15_top5_recommendations_for_ui.xlsx", index=False)
top5_df.to_parquet("15_top5_recommendations_for_ui.parquet", index=False)

print("Saved -> 15_top5_recommendations_for_ui.xlsx")
print("Saved -> 15_top5_recommendations_for_ui.parquet")

top5_df.head(15)
```
