```python
import pandas as pd

esco_rel = pd.read_csv("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/dataset/ESCO_taxonomy/esco_keep/occupationSkillRelations_en.csv")

# giữ cột cần thiết
esco_rel = esco_rel[
    [
        "occupationUri",
        "occupationLabel",
        "relationType",
        "skillType",
        "skillUri",
        "skillLabel",
    ]
].copy()

esco_rel.head()
```

# Tạo bộ từ khóa nghề IT

```python
it_patterns = [
    "ict ",
    "software",
    "web developer",
    "web designer",
    "webmaster",
    "mobile application developer",
    "cloud devops engineer",
    "cloud engineer",
    "cloud architect",
    "cloud software developer",
    "database administrator",
    "database designer",
    "database developer",
    "database integrator",
    "data engineer",
    "data scientist",
    "data analyst",
    "data warehouse designer",
    "blockchain developer",
    "blockchain architect",
    "embedded system designer",
    "embedded systems software developer",
    "embedded systems security engineer",
    "artificial intelligence engineer",
    "computer scientist",
    "computer hardware engineer",
    "computer vision engineer",
    "bioinformatics scientist",
    "software architect",
    "software developer",
    "software tester",
    "software analyst",
    "user interface developer",
    "user experience analyst",
    "telecommunications engineer",
    "telecommunications analyst",
    "network administrator",
    "network architect",
    "network engineer",
    "system administrator",
    "system architect",
    "system developer",
    "system analyst",
    "system tester",
    "test analyst",
    "integration tester",
    "qa engineer",
    "security administrator",
    "cybersecurity",
    "iot developer",
]
```

# Lọc nghề IT

```python
import re

pattern = "|".join(re.escape(x) for x in it_patterns)

esco_it_refined = esco_rel[
    esco_rel["occupationLabel"].str.lower().str.contains(pattern, na=False)
].copy()

print("ESCO IT refined:", esco_it_refined.shape)
sorted(esco_it_refined["occupationLabel"].dropna().unique())[:200]
```

# Nhóm essential skills - BẮT BUỘC CÓ

```python
essential_df = (
    esco_it_refined[esco_it_refined["relationType"].str.lower() == "essential"]
    .groupby(["occupationUri", "occupationLabel"])["skillLabel"]
    .apply(lambda x: sorted(set(x.dropna())))
    .reset_index(name="essential_skills")
)
```

# Nhóm optional skills - KHÔNG BẮT BUỘC

```python
optional_df = (
    esco_it_refined[esco_it_refined["relationType"].str.lower() == "optional"]
    .groupby(["occupationUri", "occupationLabel"])["skillLabel"]
    .apply(lambda x: sorted(set(x.dropna())))
    .reset_index(name="optional_skills")
)
```

# Gộp lại

```python
esco_it_taxonomy = essential_df.merge(
    optional_df,
    on=["occupationUri", "occupationLabel"],
    how="outer"
)
esco_it_taxonomy.sample(10, random_state=42)

esco_it_taxonomy.head()
```
