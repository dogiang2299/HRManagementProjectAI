# LƯỢT 1: LỌC AUTO

```python
import pandas as pd
import re

# ===== 1. Load ESCO occupations =====
# sửa path theo file của 
df_occ = pd.read_csv("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/dataset/ESCO_taxonomy/esco_keep/occupations_en.csv")

# giả sử cột title là preferredLabel
df_occ["title_clean"] = (
    df_occ["preferredLabel"]
    .fillna("")
    .str.lower()
    .str.strip()
)

CORE_KEYWORDS = [
    "developer", "software engineer", "programmer", "web developer",
    "frontend", "front-end", "backend", "back-end", "full stack", "full-stack",
    "mobile developer", "application developer", "embedded", "game developer",
    "data analyst", "data engineer", "data scientist", "machine learning",
    "ai engineer", "artificial intelligence", "bi developer", "business intelligence",
    "database", "dba",
    "system administrator", "systems administrator", "network administrator",
    "network engineer", "cloud engineer", "platform engineer", "infrastructure engineer",
    "site reliability", "devops", "devsecops",
    "cybersecurity", "security engineer", "information security", "penetration tester",
    "software tester", "qa engineer", "test automation", "performance tester",
    "software architect", "solution architect", "technical architect",
    "application support", "system support", "ict support"
]

EXCLUDE_KEYWORDS = [
    "sales", "marketing", "human resources", "hr", "recruiter", "accountant",
    "secretary", "office", "administrative", "customer service",
    "graphic designer", "fashion designer", "interior designer",
    "mechanical engineer", "civil engineer", "electrical engineer",
    "warehouse", "logistics", "procurement", "legal", "lawyer"
]

def contains_any(text, keywords):
    return any(k in text for k in keywords)

df_occ["is_excluded"] = df_occ["title_clean"].apply(lambda x: contains_any(x, EXCLUDE_KEYWORDS))
df_occ["is_core"] = df_occ["title_clean"].apply(lambda x: contains_any(x, CORE_KEYWORDS))

df_core = df_occ[(df_occ["is_core"]) & (~df_occ["is_excluded"])].copy()

print("Core occupations:", len(df_core))
df_core[["preferredLabel"]].drop_duplicates().sort_values("preferredLabel").head(100)
```

# Thêm bước review thủ công

```python
core_titles = (
    df_core["preferredLabel"]
    .dropna()
    .drop_duplicates()
    .sort_values()
    .tolist()
)

for t in core_titles[:200]:
    print(t)
```
