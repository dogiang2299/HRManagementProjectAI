# Cell 2 chuẩn hơn: lọc relations theo tập nghề IT

```python
import pandas as pd
import os

# ===== 1. Load merged IT occupations =====
df_occ_all = pd.read_csv(
    "/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/ESCO_taxonomy/notebook_clean/06_esco_it_all_occupations.csv"
)

# ===== 2. Load ESCO occupation-skill relations =====
df_rel = pd.read_csv(
    "/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/dataset/ESCO_taxonomy/esco_keep/occupationSkillRelations_en.csv"
)

# ===== 3. Drop unnamed columns =====
df_rel = df_rel.loc[:, ~df_rel.columns.str.contains("^Unnamed")].copy()

# ===== 4. Clean labels =====
df_occ_all["preferredLabel_clean"] = (
    df_occ_all["preferredLabel"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

df_rel["occupationLabel_clean"] = (
    df_rel["occupationLabel"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

df_rel["skillLabel_clean"] = (
    df_rel["skillLabel"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

# ===== 5. Create IT occupation set =====
it_occ_set = set(df_occ_all["preferredLabel_clean"])

# ===== 6. Filter only relations whose occupation is in IT set =====
df_rel_it = df_rel[df_rel["occupationLabel_clean"].isin(it_occ_set)].copy()

# Nếu occupationLabel_clean trong relation
# có nằm trong danh sách occupation IT đã lọc không?

# Có → giữ relation đó
# Không → bỏ

print("Total IT occupation-skill relations:", len(df_rel_it))
print("Unique IT occupations in relations:", df_rel_it["occupationLabel_clean"].nunique())
print("Unique IT skills in relations:", df_rel_it["skillLabel_clean"].nunique())

df_rel_it.head()

# Kết quả sau khi lọc relations theo nghề IT
```

# Gắn group core / extended vào relations

Kết quả:

- core: 2847

- extended: 428

Mục đích:

- nối ngược từ file nghề IT tổng để biết mỗi quan hệ thuộc nhóm nào

- giúp tách riêng relations của core và extended nếu cần

Gắn nhãn core/extended cho từng dòng quan hệ occupation-skill dựa trên nhóm nghề đã phân loại trước đó vì: File occupations đã có nhãn group, còn file occupation-skill relations 

thì chưa có, nên cần map lại nhãn core/extended từ occupation sang relation.

```python
# ===== 7. Map group from occupation file =====
occ_group_map = dict(
    zip(df_occ_all["preferredLabel_clean"], df_occ_all["group"])
)

df_rel_it["group"] = df_rel_it["occupationLabel_clean"].map(occ_group_map)

print(df_rel_it["group"].value_counts(dropna=False))
df_rel_it[["occupationLabel", "skillLabel", "group"]].head(20)

# Đây là số dòng relation thuộc từng nhóm.

# Tức là trong 3275 dòng tổng:

# 2847 dòng thuộc nhóm core

# 428 dòng thuộc nhóm extended

# Hiểu đơn giản:

# các nghề core kéo theo tổng cộng 2847 quan hệ nghề-skill

# các nghề extended kéo theo tổng cộng 428 quan hệ nghề-skill
```
