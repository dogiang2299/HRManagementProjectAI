# Thống kê số skill theo từng occupation để xem mỗi nghề IT được mô tả bởi bao nhiêu kỹ năng: mỗi nghề có bao nhiêu skill

```python
import pandas as pd

df_rel_it = pd.read_csv(
    "/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/ESCO_taxonomy/notebook_clean/07_esco_it_occupation_skill_relations.csv"
)

# Làm sạch nhẹ cho chắc
df_rel_it["occupationLabel_clean"] = (
    df_rel_it["occupationLabel"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

df_rel_it["skillLabel_clean"] = (
    df_rel_it["skillLabel"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

# Đếm số skill khác nhau theo từng occupation
occ_skill_count = (
    df_rel_it.groupby(["occupationLabel", "group"])["skillLabel_clean"]
    .nunique()
    .reset_index(name="num_skills")
    .sort_values("num_skills", ascending=False)
    .reset_index(drop=True)
)

print("Number of occupations:", len(occ_skill_count))
occ_skill_count.head(20)
```

```python
occ_skill_count.to_csv(
    "/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/ESCO_taxonomy/notebook_clean/08_esco_occupation_skill_count.csv",
    index=False
)

print("Saved occupation skill count summary.")
```
