```python
import pandas as pd

# ===== 1) ĐỌC FILE =====
df = pd.read_excel("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/Djinni/notebook_clean/03_djinni_market_alignment.xlsx")

# ===== 2) HÀM DỌN CHỮ =====
def lam_sach_chu(text):
    if pd.isna(text):
        return ""
    text = str(text).strip().lower()
    text = text.replace("\r\n", " ").replace("\n", " ").replace("\r", " ")
    text = " ".join(text.split())
    return text

# ===== 3) ĐẢM BẢO CÓ CỘT CẦN DÙNG =====
for col in ["ten_sach", "ten_thi_truong", "ten_gan_giong", "huong_xu_ly"]:
    if col not in df.columns:
        df[col] = ""

df["ten_sach"] = df["ten_sach"].apply(lam_sach_chu)
df["ten_thi_truong"] = df["ten_thi_truong"].fillna("")
df["ten_gan_giong"] = df["ten_gan_giong"].fillna("")
df["huong_xu_ly"] = df["huong_xu_ly"].fillna("")

# ===== 4) BẢNG GỢI Ý NHANH =====
goi_y = {
    "frontend development": {
        "ten_thi_truong": "frontend",
        "ten_gan_giong": "front-end, front end, fe",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "backend development": {
        "ten_thi_truong": "backend",
        "ten_gan_giong": "back-end, back end, be",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "fullstack development": {
        "ten_thi_truong": "fullstack",
        "ten_gan_giong": "full-stack, full stack",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "quality assurance": {
        "ten_thi_truong": "qa",
        "ten_gan_giong": "tester, software testing, manual testing, qc",
        "huong_xu_ly": "doi_ten"
    },
    "software testing": {
        "ten_thi_truong": "testing",
        "ten_gan_giong": "qa, tester, manual testing",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "devops": {
        "ten_thi_truong": "devops",
        "ten_gan_giong": "sre, ci/cd, platform engineering",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "system administration": {
        "ten_thi_truong": "system admin",
        "ten_gan_giong": "sysadmin, server administration",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "database administration": {
        "ten_thi_truong": "dba",
        "ten_gan_giong": "database admin, database administrator",
        "huong_xu_ly": "doi_ten"
    },
    "data engineering": {
        "ten_thi_truong": "data engineer",
        "ten_gan_giong": "etl, elt, data pipeline, big data",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "data analysis": {
        "ten_thi_truong": "data analyst",
        "ten_gan_giong": "analytics, bi, business intelligence",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "machine learning": {
        "ten_thi_truong": "machine learning",
        "ten_gan_giong": "ml, ai, model training",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "cybersecurity": {
        "ten_thi_truong": "cybersecurity",
        "ten_gan_giong": "security, infosec, information security",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "technical support": {
        "ten_thi_truong": "it support",
        "ten_gan_giong": "helpdesk, service desk, technical support",
        "huong_xu_ly": "doi_ten"
    },
    "ui design": {
        "ten_thi_truong": "ui design",
        "ten_gan_giong": "user interface design, ui",
        "huong_xu_ly": "them_ten_gan_giong"
    },
    "ux design": {
        "ten_thi_truong": "ux design",
        "ten_gan_giong": "user experience design, ux",
        "huong_xu_ly": "them_ten_gan_giong"
    },
}

# ===== 5) DANH SÁCH TỪ NÊN GIỮ NGUYÊN =====
giu_nguyen = {
    "python", "java", "javascript", "typescript", "php", "ruby", "go", "golang",
    "c", "c++", "c#", "dotnet", ".net", "react", "angular", "vue", "nodejs",
    "nestjs", "spring", "django", "flask", "laravel",
    "sql", "mysql", "postgresql", "mongodb", "redis",
    "docker", "kubernetes", "git", "linux", "aws", "azure", "gcp",
    "html", "css", "sass", "figma", "jira", "graphql", "rest api"
}

# ===== 6) DANH SÁCH TÊN MƠ HỒ NÊN XEM LẠI =====
xem_lai = {
    "security", "management", "architecture", "support",
    "engineering", "testing", "development", "design", "analysis", "data"
}

# ===== 7) HÀM ĐIỀN TỰ ĐỘNG =====
def dien_tu_dong(row):
    ten = row["ten_sach"]

    # nếu đã điền rồi thì giữ nguyên
    if str(row["huong_xu_ly"]).strip() != "":
        return row

    if ten in goi_y:
        row["ten_thi_truong"] = goi_y[ten]["ten_thi_truong"]
        row["ten_gan_giong"] = goi_y[ten]["ten_gan_giong"]
        row["huong_xu_ly"] = goi_y[ten]["huong_xu_ly"]
        return row

    if ten in giu_nguyen:
        row["ten_thi_truong"] = ten
        row["ten_gan_giong"] = ""
        row["huong_xu_ly"] = "giu_nguyen"
        return row

    if ten in xem_lai:
        row["ten_thi_truong"] = ""
        row["ten_gan_giong"] = ""
        row["huong_xu_ly"] = "xem_lai"
        return row

    # một số luật theo từ chứa bên trong
    if "frontend" in ten:
        row["ten_thi_truong"] = "frontend"
        row["ten_gan_giong"] = "front-end, front end, fe"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "backend" in ten:
        row["ten_thi_truong"] = "backend"
        row["ten_gan_giong"] = "back-end, back end, be"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "qa" in ten or "quality assurance" in ten:
        row["ten_thi_truong"] = "qa"
        row["ten_gan_giong"] = "tester, software testing, qc"
        row["huong_xu_ly"] = "doi_ten"
        return row

    if "test" in ten:
        row["ten_thi_truong"] = "testing"
        row["ten_gan_giong"] = "qa, tester, manual testing"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "devops" in ten:
        row["ten_thi_truong"] = "devops"
        row["ten_gan_giong"] = "sre, ci/cd, platform engineering"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "data engineer" in ten or "data engineering" in ten:
        row["ten_thi_truong"] = "data engineer"
        row["ten_gan_giong"] = "etl, elt, data pipeline"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "data analyst" in ten or "data analysis" in ten:
        row["ten_thi_truong"] = "data analyst"
        row["ten_gan_giong"] = "analytics, bi, business intelligence"
        row["huong_xu_ly"] = "them_ten_gan_giong"
        return row

    if "security" in ten:
        row["ten_thi_truong"] = "security"
        row["ten_gan_giong"] = "cybersecurity, infosec, information security"
        row["huong_xu_ly"] = "xem_lai"
        return row

    return row

# ===== 8) ÁP DỤNG =====
df = df.apply(dien_tu_dong, axis=1)

# ===== 9) LƯU FILE =====
df.to_excel("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/Djinni/notebook_clean/04_djinni_round1_step1.xlsx", index=False)

print("Đã tạo file: 04_djinni_round1_step1.xlsx")
print(df["huong_xu_ly"].value_counts(dropna=False))
```
