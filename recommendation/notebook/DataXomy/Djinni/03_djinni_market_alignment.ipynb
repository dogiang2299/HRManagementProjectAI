```python
import pandas as pd

# ===== 1) ĐỌC FILE 14 CLEAN =====
df = pd.read_excel("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/ESCO_taxonomy/notebook_clean/14_skill_mapping_clean.xlsx")

# ===== 2) HÀM DỌN CHỮ =====
def lam_sach_chu(text):
    if pd.isna(text):
        return ""
    text = str(text)
    text = text.replace("\r\n", " ")
    text = text.replace("\n", " ")
    text = text.replace("\r", " ")
    text = " ".join(text.split())
    return text.strip()

# ===== 3) HÀM CHUẨN HOÁ TÊN =====
def chuan_hoa_ten(text):
    text = lam_sach_chu(text).lower()

    # gom các cách viết khác nhau về một kiểu
    text = text.replace("front-end", "frontend")
    text = text.replace("front end", "frontend")
    text = text.replace("back-end", "backend")
    text = text.replace("back end", "backend")
    text = text.replace("full-stack", "fullstack")
    text = text.replace("full stack", "fullstack")
    text = text.replace("node.js", "nodejs")
    text = text.replace(".net", "dotnet")

    return text.strip()

# ===== 4) ĐỔI TÊN CỘT CHO DỄ LÀM =====
df = df.rename(columns={
    "mapped_taxonomy_group": "nhom_lon",
    "mapped_taxonomy_subgroup": "nhom_nho",
    "skill_group": "cum_ky_nang",
    "skill_name": "ten_goc"
})

# ===== 5) GIỮ LẠI CÁC CỘT CẦN CHO PHA A =====
cot_can_lay = [
    "nhom_lon",
    "nhom_nho",
    "cum_ky_nang",
    "ten_goc",
    "skill_id",
    "skill_type",
    "group",
    "relation_count",
    "skill_subgroup",
    "notes"
]

df = df[cot_can_lay].copy()

# ===== 6) LÀM SẠCH CỘT TÊN =====
df["ten_goc"] = df["ten_goc"].apply(lam_sach_chu)
df["ten_sach"] = df["ten_goc"].apply(chuan_hoa_ten)

# ===== 7) THÊM CÁC CỘT CHO DJINNI ROUND 1 - BƯỚC 1 =====
df["ten_thi_truong"] = ""
df["ten_gan_giong"] = ""
df["huong_xu_ly"] = ""
df["ghi_chu_djinni"] = ""
df["da_kiem"] = 0
df["vong"] = 1
df["buoc"] = 1

# ===== 8) SẮP XẾP CHO DỄ NHÌN =====
thu_tu_cot = [
    "nhom_lon",
    "nhom_nho",
    "cum_ky_nang",
    "skill_subgroup",
    "ten_goc",
    "ten_sach",
    "ten_thi_truong",
    "ten_gan_giong",
    "huong_xu_ly",
    "ghi_chu_djinni",
    "skill_id",
    "skill_type",
    "group",
    "relation_count",
    "notes",
    "da_kiem",
    "vong",
    "buoc"
]

df = df[thu_tu_cot]

# ===== 9) LƯU FILE MỚI =====
df.to_excel("/Users/nguyenduykhanh/Documents/GraduationProject/Recommendation_Project/recommendation/notebook/Djinni/notebook_clean/03_djinni_market_alignment.xlsx", sheet_name="djinni_round1", index=False)

print("Đã tạo xong file: 03_djinni_market_alignment.xlsx")
print("Số dòng:", len(df))
print("Số cột:", len(df.columns))
```
